```python
import numpy as np
```

```python
#Math Matical Operation on single Array
```

```python
a = np.array([10,30,20,-11,2])
print(a)
```

```python
print(a+13)
print(a-3)
print(a*4)
print(a/2)
```

```python
b = np.array([121,9,4])
print(np.sqrt(b))
print(np.square(b))
```

```python
#Mathmetical Operation on multiple arrays
```

```python
a1 = np.array([1,2,3,4])
b1 = np.array([5,6,7,8])
print(a1)
print(b1)
```

```python
print(np.add(a1,b1))
print(np.subtract(a1,b1))
print(np.multiply(a1,b1))
print(np.divide(a1,b1))
```

```python
print(np.dot(a1,b1))
```

```python
print(a1.T)
```
