```python
import numpy as np
```

```python
#Broadcasting
a = np.array([1,2,3,4,5])
b = np.array([6])
```

```python
print(a+b)
```

```python
c = np.array([[1,2,3,4], [5,6,7,8]])
d = np.array([3,7,6,9])
```

```python
print(c+d)
print(c-d)
print(c*d)
print(c/d)
print(c%d)
```
