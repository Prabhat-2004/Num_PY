```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Type:", type(arr))
print("D-Type:", arr.dtype)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Dimension:", arr.ndim)
print("Data Type:", arr.dtype)
```

```python
#List to array coversion
import numpy as np
lst1 = [11.0,9,8,7,3.2]
lst2 = [6,5,4,3,2.1]
lst3 = [2,1,6,0,4.4]
print(lst1,lst2,lst3)
arr_one_dimension = np.array(lst1)
```

```python
#One Dimensional Array
print(arr_one_dimension)
print("DTYPE:", arr_one_dimension.dtype)
```

```python
#Two Dimensional Array
arr_two_dimension = np.array([lst1,lst2])
print(arr_two_dimension)
```

```python
#Three Dimensional Array
arr_three_dimension = np.array([[lst1,lst2,lst3]])
print(arr_three_dimension)
```
