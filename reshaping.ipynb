```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5,6])
print(arr)
```

```python
reshape_arr = arr.reshape((6,1))
print(reshape_arr)
```

```python
reshape_arr2 = arr.reshape((1,6))
print(reshape_arr2)
```

```python
reshape_arr3 = arr.reshape((2,3))
print(reshape_arr3)
```

```python
reshape_arr4 = arr.reshape((3,1,2))
print(reshape_arr4)
```

```python
flatten_arr5 = reshape_arr4.flatten()
print(flatten_arr5)
```
